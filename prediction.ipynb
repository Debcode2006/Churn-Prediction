```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import numpy as np
import pandas as pd
import pickle
```

```python
model = load_model('model.h5')
```

```python
def pickle_loader(path:str):
    with open (path,"rb") as file:
        return pickle.load(file)
    
label_encoder_gender=pickle_loader("label_encoder_gender.pkl")
ohe_geography=pickle_loader("onehot_encoder_geography.pkl")
scaler=pickle_loader("scaler.pkl")
```

```python
input_data={
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
```

```python
ohe_geography.get_feature_names_out(['Geography'])
```

```python
geo_encoded = ohe_geography.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded,columns=ohe_geography.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_data_df=pd.DataFrame([input_data])
```

```python
input_data_df=pd.concat([input_data_df.reset_index(drop=True),geo_encoded_df],axis=1)
```

```python
input_data_df['Gender']=label_encoder_gender.transform(input_data_df['Gender'])
```

```python
input_data_df.drop(['Geography'],axis=1,inplace=True)
```

```python
input_scaled=scaler.transform(input_data_df)
```

```python
prediction=model.predict(input_scaled)
if prediction>0.5:
    print("Customer will churn")
else:
    print("Customer not likely to churn")
```
